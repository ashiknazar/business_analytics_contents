#### 1. Project Background (What the Intern Must Understand)

- Your company is hired by ShopEase Retail, an online retailer that is experiencing:
  - Excess inventory in some warehouses
  - Frequent stockouts in others
  - High logistics costs

- The client wants a BA team to understand the operational problems and design a solution roadmap.
- Goal: Teach the intern how a BA works in a real client engagement from start to finish.

___

### 2. Step 1: Client Communication (What a BA Says and Does)

- The BA intern must learn how to initiate conversations, ask questions, and document outcomes.
- Below is the actual communication flow used in industry.


#### 2.1 The BA Schedules a Requirements Workshop

- People invited:
- Client Operations Manager
- Warehouse Lead
- E-commerce Product Manager
- Finance Analyst

#### BA Opening Statement:
"Thank you for joining. Our objective today is to understand your operational challenges in the inventory lifecycle so we can define the problem accurately and propose a data-driven business solution."

### 2.2 Critical Questions the BA Must Ask

A business analyst intern must be trained to ask questions in these domains:

#### a. Problem Understanding
- What business challenges are you experiencing?
- How is inventory currently monitored and replenished?
- Which KPIs are negatively affected?
#### b. Current Process Understanding
- What is your process for forecasting demand?
- How are vendor orders raised?
- Which systems are used (ERP, WMS, spreadsheets)?

#### c. Pain Points
- Which SKUs usually overstock?
- Where do stockouts occur?
- Are costs rising due to emergency replenishments?

#### d. Data Availability
- What data do you currently collect?
- What data is missing that could help?
- Do you track forecast error? Lead time? Vendor quality?
#### e. Success Criteria
- What outcome will make this project successful?
(e.g., 15% reduction in excess inventory)

### 2.3 BA Creates "Requirements Elicitation Notes"

Example:

- Client Stated Problems:
  - Overstocking for slow-moving products
  - Stockouts during peak season
  - Poor demand forecasting
  - Manual tracking of inventory levels

- Constraints:
  - Data is split across multiple systems
  - Warehouse staff manually enters product movement once a day

- Risks Identified:
  - Data quality may be inconsistent
  - Change resistance from warehouse staff

### 3. Step 2: Requirements Documentation (Intern Learns Real Industry Templates)

A BA intern must be able to prepare:

##### 1. BRD (Business Requirements Document)

- Contains:
  - Background
  - Business problem
  - Stakeholders
  - High-level requirements
  - KPIs
  - Constraints

##### 2. FRD (Functional Requirements Document)

- Contains:
  - What the new solution should do
  - Process changes
  - System behavior
  - Data fields needed
  - User roles

##### 3. Process Maps
  - AS-IS process
  - TO-BE process
  
Using BPMN diagrams.

##### 4. User Stories

Example:
- "As an inventory planner, I want automated weekly demand forecasts so that I can plan procurement efficiently."

### 4. Step 3: Data Collection (What a BA Requests)

The BA must request the minimum necessary data from the client.

| Data Type         | Description                            |
| ----------------- | -------------------------------------- |
| Sales History     | SKU-level sales per day/week           |
| Inventory Logs    | Beginning stock, receipts, adjustments |
| Forecast data     | Existing forecasts (if any)            |
| Lead Time         | Vendor-wise lead time                  |
| Warehouse Data    | Location, capacity, constraints        |
| Seasonality Flags | Events, promotions, holidays           |


##### 4.1 BA Asks If Additional Data Can Be Generated

This is important for training the intern.
##### BA Recommendation Examples:
- Introduce Stock Movement Granularity (hourly logs)
- Capture Lost Sales (pages viewed but item out of stock)
- Introduce Demand Classification (fast, medium, slow)
- Track Delivery Date Variability from vendors

Intern must learn how to recommend new data fields when needed.

#### 5. Step 4: Business Analysis (Intern Executes the Analysis)

A BA is not a data scientist—but must perform insight-level analysis.

Key analysis tasks:

##### 5.1 KPI Analysis
- Current inventory turnover ratio
- Number of stockouts per warehouse
- Days of inventory on hand (DOH)
- Demand variability per SKU

##### 5.2 Trend Analysis
- Slow-moving vs fast-moving products
- Seasonal demand spikes
- Warehouse-level performance differences

##### 5.3 Process Gaps
- Identifying manual steps causing delays
- Redundant approvals in procurement process
- Lack of automated alerts

##### 5.4 Root Cause Analysis
Tools:
 - Fishbone diagram
 - 5 Whys
 - SIPOC diagram

Example Finding:
Overstock happens because forecasting is done manually using past 2 weeks of sales, ignoring seasonality.

#### 6. Step 5: Building Recommendations (BA Output)

The intern must prepare actionable recommendations such as:

##### Operational Improvements
- Introduce ABC classification for inventory
- Standardize replenishment reorder points
- Reduce manual entry delays by integrating WMS with ERP

##### Technology Recommendations
 - Introduce a simple forecasting module (not ML; could be rules-based)
 - Automated stock alerts to procurement team

##### Process Improvements
- Weekly S&OP meeting
- Vendor scorecard for lead time accuracy

#### 7. Step 6: Reporting & Presentations

A BA intern must deliver:

##### A. Business Analysis Report

Sections:
- Executive summary
- Identified problems
- KPI findings
- Root cause insights
- Recommendations
- Expected benefits

#### B. Slides for Client Review
- Pain points
- AS-IS vs TO-BE process
- Quantified impact
- Roadmap

##### C. Requirements Documentation (BRD + FRD)
##### D. Process diagrams (BPMN)
##### E. User Stories & Acceptance Criteria
