# Financial Services – Industry Breakdown

## 1. Banking
### Retail Banking
- Products: Savings accounts, credit cards, personal loans, mortgages
- Typical Problems:
  - Credit risk modeling (PD, LGD, EAD)
  - Customer churn prediction
  - Cross-sell / upsell modeling
  - Fraud detection

### Corporate Banking
- Products: Business loans, trade finance, treasury services
- Typical Problems:
  - Counterparty risk assessment
  - Cash flow forecasting
  - Loan default prediction
  - ESG risk scoring

### Investment Banking
- Services: M&A advisory, IPO underwriting, structured products
- Typical Problems:
  - Deal valuation modeling
  - Market risk (VaR, CVaR)
  - Liquidity stress testing
  - Scenario simulation

---

## 2. FinTech
- Segments: Digital payments, BNPL, Neo-banks, Lending platforms
- Typical Problems:
  - Real-time fraud detection
  - Alternative credit scoring
  - KYC automation (OCR + NLP)
  - Transaction anomaly detection

---

## 3. Insurance
- Segments: Life, Health, General (Auto, Property), Reinsurance
- Typical Problems:
  - Underwriting risk scoring
  - Claim fraud detection
  - Loss reserving prediction
  - Customer lifetime value modeling

---

## 4. Asset Management
- Segments: Mutual Funds, Hedge Funds, Pension Funds, Private Equity
- Typical Problems:
  - Portfolio optimization
  - Alpha generation models
  - Factor modeling
  - Risk parity allocation

---

## 5. Stock Markets & Trading
- Segments: Exchanges, Brokers, HFT firms, Market makers
- Typical Problems:
  - Algorithmic trading strategies
  - Order book prediction
  - Market microstructure analysis
  - Volatility forecasting
