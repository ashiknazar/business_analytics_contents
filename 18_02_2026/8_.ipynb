#### Tools

##### 1. Spreadsheet & Basic Analysis Tools

- Microsoft Excel
- Type: Spreadsheet & quick analytics tool
  - Fast exploratory analysis
  - Quick KPI tracking
  - Financial modeling
  - Scenario planning
- What intern should master:
  - Pivot Tables
  - XLOOKUP / INDEX-MATCH
  - Conditional formatting
  - What-if analysis
  - Basic dashboards
  - Solver (for optimization basics)

**Business Use Case:**

Monthly sales reporting, budgeting, profit analysis.

____

##### 2. Database & Querying Tools

**SQL (MySQL / PostgreSQL)**

Type: Data querying language

- Why it is used:
  - Extract data from databases
  - Perform aggregations
  - Create business reports
  - Work with data warehouses
- What intern should master:
  - SELECT, WHERE, GROUP BY
  - JOINs
  - Window functions
  - CTEs
  - Aggregation logic 
  - Writing optimized queries

**Business Use Case:**

Customer segmentation, revenue analysis, operational reporting.

___

## 3. Programming for Analytics

### Python  

**Type:**  
Data Analysis & Predictive Modeling Tool  

---

### Libraries Used

- **Pandas** – Data manipulation and transformation  
- **NumPy** – Numerical computing and array operations  
- **Matplotlib / Seaborn** – Data visualization  
- **Scikit-learn** – Machine learning modeling  
- **Statsmodels** – Statistical analysis and hypothesis testing  

---

### Why It Is Used

- Enables advanced analytics beyond Excel  
- Automates reporting workflows  
- Supports predictive modeling  
- Efficiently handles large datasets  

---

### What an Intern Should Master

- Data cleaning and preprocessing  
- Exploratory Data Analysis (EDA)  
- Hypothesis testing  
- Regression and classification models  
- Model evaluation metrics  

---

### Business Use Cases

- Customer churn prediction  
- Demand forecasting  
- A/B test analysis
