#### Executive summary (one paragraph)
- Business analytics is the application of statistical methods, quantitative models, and structured problem-solving to convert operational data into actionable business decisions. It sits between raw data work (data engineering/analytics) and business strategy, combining analytical outputs with requirements, process design, stakeholder management, and decision frameworks to produce business outcomes.

### 1 — Foundational theories and principles

- These are the intellectual foundations interns must understand and be able to apply.

##### 1.1 Decision theory and value of information

- **Core idea:** Decisions under uncertainty require comparing expected outcomes; information that reduces uncertainty has measurable value. Use expected value calculations and decision trees to justify data collection or experiments.

- **How to teach**: build a decision tree for “invest in marketing campaign vs. not” with probabilities and expected ROI(return on investment); compute value of perfect information (VPI) for a costly survey.

##### 1.2 Statistical inference & sampling theory

- **Core idea:** Estimates from samples must include measures of uncertainty (confidence intervals, p-values). Misuse yields wrong business forks.

- **Teaching:** hypothesis tests for lift in A/B tests; sample-size calculator for detecting 5% lift at 80% power.

##### 1.3 Causal thinking and experimental design

- **Core idea:** Correlation ≠ causation. Randomized controlled trials (A/B tests), difference-in-differences, and instrumental variables are tools to estimate causal effects.

- **Teaching:** design an A/B test for checkout page change; compute treatment/control size, check for bias, run t-tests/bootstrapping.

##### 1.4 Predictive modeling theory (bias–variance, regularization)

- **Core idea:** Tradeoffs between model complexity and generalization; regularization (L1/L2) combats overfitting; interpretability vs. performance tradeoffs must be managed for business acceptance.

- **Teaching:** fit linear, ridge, and lasso models on the same dataset and compare RMSE and coefficient stability.

##### 1.5 Time-series & forecasting theory

- **Core idea:** Forecasts must model trend, seasonality and noise; evaluate using MAPE, RMSE, and coverage for prediction intervals. Use ARIMA/ETS/Prophet for different data properties.
- **Teaching:** forecast weekly demand for 2 stores; decompose series; compare simple exponential smoothing vs. Prophet.

### 2 — Core techniques & methods (practical toolkit)

- These are the tactical methods every business analyst must execute reliably.

##### 2.1 Problem framing & hypothesis articulation

- Convert a business question into measurable hypotheses and KPIs (e.g., increase conversion by X% in Y weeks).

- Deliverable: one-page problem statement with KPIs, data needs, stakeholders.

##### 2.2 Exploratory Data Analysis (EDA)

Profiling: distributions, missingness, cardinality, time ranges.

Segmentation: cohort analysis, RFM, cluster-based segments.

Visualization: histograms, boxplots, cumulative curves (Lorenz), funnel plots.

##### 2.3 Data quality assessment & remediation

- **Metrics:** completeness, consistency, uniqueness, timeliness, accuracy.
- **Methods:** deduplication, canonicalization, validation rules, audit logs.
- **Deliverable:** Data quality dashboard + remediation plan.

##### 2.4 Reporting and dashboarding
- KPI definition, drill-down capability, operational alerts.
- **Template:** Executive page (summary KPIs), Operational page (daily metrics), Root-cause page (filters + cohort analysis).

##### 2.5 Requirements elicitation & documentation

- BRD/FRD, user stories, acceptance criteria, use-case diagrams, RTM (requirement traceability matrix). Practical templates from Asana/Smartsheet/Canva are industry standard.

##### 2.6 Stakeholder communication & storyboarding
- Convert analytical findings into a 5–7 slide “insight + action” deck: context, findings, impact (quantified), recommended actions, next steps.

##### 2.7 Predictive modeling & scoring
- **Common algorithms:** logistic regression, decision trees, random forest, gradient boosting, simple neural nets. Include interpretability tools (SHAP/LIME) to explain predictions.
- **Deliverable:** model card (purpose, data, features, validation results, fairness checks).

##### 2.8 Experimentation & A/B testing
- Randomization strategy, metric hierarchy (primary/secondary), sequential monitoring rules, sample size estimation.
- Deliverable: A/B test plan + analysis report with confidence intervals and business recommendation.

##### 2.9 Optimization & decision analytics
- Linear programming for resource allocation, integer programming for scheduling, simple simulation for “what if” analysis.
- Deliverable: optimization model + sensitivity analysis.

##### 2.10 Forecasting & inventory/demand planning

- Short horizon vs long horizon approaches, intermittent demand methods, error measurement, backtesting. Use rolling-origin evaluation.

### 3 — Essential skills (non-technical + technical)

- Organize interns’ competency map into hard and soft skills.

##### Technical skills

- Data literacy: SQL fluency (joins, windows, aggregations), data modeling basics.
- Analysis: Excel (pivot tables, power pivot), Python (pandas, NumPy), basic statistics.
- Visualization: Power BI or Tableau proficiency; ability to build executive dashboards.
- Basic ML: model training/validation, feature engineering, baseline models, interpretation.

##### Business/soft skills
- Problem scoping & stakeholder interviews (elicitation).
- Documentation: BRD, use cases, acceptance criteria. Templates: Asana/Smartsheet/Canva. 
- Presentation & persuasion: quantify impact and recommended decisions.
- Process mapping: BPMN/flowcharts, SIPOC.

##### 4 — Tools & technology stack (recommended)


##### Data access & storage
- SQL databases: PostgreSQL, MySQL, BigQuery (for analytics).
- NoSQL/warehouse: Redshift, Snowflake (enterprise scale).

##### Data wrangling & scripting
- Excel (for quick prototyping), Python (pandas), R (optional).

##### Modeling & ML
- scikit-learn, XGBoost, LightGBM, Prophet for time series. Use SHAP for interpretability. (See case studies referenced below.)

##### Visualization & dashboards
- Tableau, Power BI, Looker (choose one primary tool for consistency). Many companies use Power BI or Tableau as standard.

##### Collaboration & documentation
- Jira + Confluence (Agile BA workflows), Miro/Lucidchart for process diagrams, Figma for wireframes. BRD/FRD templates available from Asana/Smartsheet/Canva.

##### 5 — Deliverables & templates (intern-ready)

Provide explicit deliverables so interns know what “done” looks like.

- One-page Problem Statement (context, KPI, stakeholders, data sources)
- Data Profile Report (counts, missingness, top values, anomalies)
- BRD / User Stories (with acceptance criteria) — use Smartsheet/Asana templates.
- EDA Notebook or Excel Workbook (scripts + narrative)
- Dashboard (Executive + Operational pages)
- Model Card (algorithm, metrics, training data, evaluation, limitations)
- Insight Deck (5–10 slides)
- Handover document (deployment steps, monitoring metrics)

##### 6 — Eight evidence-based use cases with data, approach, and expected KPIs
- Below are practical, high-value BA use cases, with method, primary data needs, evaluation metrics, and references to published work or industry templates where applicable.

## Use case A — Customer churn reduction (telecom / SaaS / banking) 

- Business goal: Reduce churn by targeted interventions that maximize retention ROI.
- Data: customer transactions, product usage, service tickets, billing, tenure, demographics.
- Approach: EDA → churn score model (logistic / gradient boosting) → segment by risk & value → design retention offers → A/B test offers.
- KPIs: churn rate, retention lift, cost per retained customer, LTV increase.
- Evidence: published churn prediction studies show high ROC AUC when combining behavioral and billing features; academic and industry case studies report substantial cost savings when risk scoring is applied.

## Use case B — Sales forecasting & inventory optimization (retail)

- Business goal: Reduce stockouts and overstock, improve fill rate.
- Data: POS transactions, promotions calendar, lead times, SKU hierarchy.
- Approach: time-series decomposition, intermittent demand models for slow movers, safety stock optimization via simulation.
- KPIs: forecast error (MAPE), stockout rate, inventory turnover.
- Evidence: Industry practice uses ARIMA/ETS/Prophet and ML hybrids; retailers report measurable reductions in stockouts with ML-enriched forecasts.

## Use case C — Marketing attribution & campaign optimization (ecommerce)
- Business goal: Allocate marketing spend to channels with highest incremental ROI.
- Data: click & impression logs, conversions, channel spend, customer journey data.
- Approach: multi-touch attribution or uplift modeling; controlled experiments for causal lift.
- KPIs: incremental conversions, ROAS (return on ad spend), CAC.
- Evidence: Experimental designs (holdout groups) provide the most credible lift estimates vs. naive attribution; HBR/business analytics texts recommend experimentation when feasible.

## Use case D — Pricing optimization
- Business goal: Maximize revenue/profit via dynamic pricing or price sensitivity analysis.
- Data: sales, price history, competitor prices, seasonality.
- Approach: demand curve estimation (price elasticity), discrete choice models, A/B pricing experiments.
- KPIs: revenue, margin, conversion rate, price elasticity coefficient.

## Use case E — Fraud detection (finance, payments)
- Business goal: Reduce fraud losses while keeping false positives (customer friction) low.
- Data: transactions, device fingerprints, historical fraud labels.
- Approach: supervised models with class imbalance techniques; rule-based systems for real-time blocking; continuous model monitoring.
- KPIs: precision@k, recall on fraudulent cases, false positive rate, dollars saved.

## Use case F — Process optimization (operations)
- Business goal: Reduce cycle time and costs (e.g., order-to-fulfill).
- Data: timestamps across process steps, resource utilization.
- Approach: process mapping (BPMN), bottleneck identification, queueing models, discrete-event simulation, optimization (LP/IP).
- KPIs: lead time, throughput, utilization, cost per order.

## Use case G — Customer segmentation & CLTV (marketing / product)
- Business goal: Identify high-value customers and tailor offers.
- Data: purchase history, recency/frequency/monetary (RFM), demographics.
- Approach: RFM scoring, clustering, predictive CLTV models.
- KPIs: CLTV, retention rate, uplift in cross-sell conversions.

##### Use case H — Workforce analytics (HR)
- Business goal: Reduce attrition in critical roles and optimize hiring pipeline.
- Data: HRIS records, performance ratings, exit interviews, engagement surveys.
- Approach: EDA, attrition predictive modeling, root-cause categorical analysis, targeted retention interventions.
- KPIs: attrition rate, time-to-fill, hiring cost, retention lift.

## 7 — Example end-to-end BA task (concrete intern assignment)

Problem: Reduce monthly active user (MAU) churn by 10% over 3 months for a subscription app.

### Steps for intern:

- Problem statement and KPIs (MAU churn, retention rate).
- Data profile: user events, billing, support tickets (one-pager)
- EDA: churn by cohort, tenure, feature usage. Visualize top drivers.
- Stakeholder interview summary (mock): list requirements and constraints.
- Build churn score (baseline logistic regression). Report: ROC-AUC, PR curve, calibration.
- Segment users into risk/value buckets; design two retention offers.
- Run holdout experiment (design plan), including sample size and metric hierarchy.
- Dashboard: weekly cohort retention and test progress.
- Final deck: findings, recommended prioritized actions, expected impact and next steps.

Deliverables: Problem statement, EDA notebook, BRD for experiment, model card, dashboard link, insight deck.
